**Segmentation for Annocerv

```python
import cv2
import numpy as np
import os

def get_tz_from_colposcopy(image):
    # Step (b): Extract the red channel
    red_channel = image[:, :, 2]

    # Step (c): Thresholding to detect the central part
    _, binary_mask = cv2.threshold(red_channel, 50, 255, cv2.THRESH_BINARY)

    # Step (d): Detect the big object using area filter
    contours, _ = cv2.findContours(binary_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    areas = [cv2.contourArea(contour) for contour in contours]
    if not areas:
        return image  # Return original if nothing found

    max_area_index = np.argmax(areas)
    big_object_mask = np.zeros_like(binary_mask)
    cv2.drawContours(big_object_mask, contours, max_area_index, 255, -1)

    # Step (e): Trace the boundary of the big object
    boundary = cv2.findContours(big_object_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)[0]
    boundary_mask = np.zeros_like(binary_mask)
    cv2.drawContours(boundary_mask, boundary, 0, 255, -1)

    # Step (f): Multiply the binary mask with the original RGB image
    tz_image = cv2.bitwise_and(image, image, mask=boundary_mask)

    return tz_image

if __name__ == "__main__":
    image_paths = []

    rootdir = r"E:\My PHD\4-Cervical research paper\4-Datasets\AnnoCerv-main\Augmented and preprocessed\case 0-4 image - Copy"
    for subdir, dirs, files in os.walk(rootdir):
        for file in files:
            if file.lower().endswith((".jpg", ".png", ".jpeg")):  # Only process image files
                full_path = os.path.join(subdir, file)
                image_paths.append(full_path)

    count = 0
    for path in image_paths:
        rgb_image = cv2.imread(path)
        if rgb_image is None:
            print(f"Failed to read {path}")
            continue

        tz_image = get_tz_from_colposcopy(rgb_image)

        # Option 1: Overwrite original
        # cv2.imwrite(path, tz_image)

        # Option 2: Save with '_tz' suffix
        base, ext = os.path.splitext(path)
        output_path = base + ext
        cv2.imwrite(output_path, tz_image)

        print(f"Saved: {output_path}")
        count += 1

    print(f"Total images processed: {count}")
```

```python
from PIL import Image
import os
import random
import math

def image_augmentation(input_folder, output_folder, total_target_images=311):
    # Create the output folder if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # List of augmentation functions
    augmentation_functions = [
        flip_image_left_right,
        flip_image_top_bottom,
        rotate_image_90,
        rotate_image_270
    ]

    # Get list of all images
    original_files = [f for f in os.listdir(input_folder) if os.path.isfile(os.path.join(input_folder, f))]
    original_count = len(original_files)
    required_augmented_count = total_target_images - original_count

    print(f"Original images: {original_count}")
    print(f"Need to generate: {required_augmented_count} augmented images")

    # Keep track of how many we’ve added
    augmented_images_created = 0

    # Shuffle the original images for fair augmentation
    random.shuffle(original_files)

    while augmented_images_created < required_augmented_count:
        for filename in original_files:
            if augmented_images_created >= required_augmented_count:
                break

            img_path = os.path.join(input_folder, filename)
            img = Image.open(img_path)

            # Randomly pick an augmentation
            aug_fn = random.choice(augmentation_functions)
            aug_img = aug_fn(img.copy())

            # Save the augmented image
            base_name = os.path.splitext(filename)[0]
            output_name = f"{base_name}_aug{augmented_images_created}.jpg"
            output_path = os.path.join(output_folder, output_name)
            aug_img.save(output_path)

            augmented_images_created += 1

    print(f"✅ Augmentation completed. Total images now = {original_count + augmented_images_created}")

# Augmentation functions
def flip_image_left_right(image):
    return image.transpose(Image.FLIP_LEFT_RIGHT)

def flip_image_top_bottom(image):
    return image.transpose(Image.FLIP_TOP_BOTTOM)

def rotate_image_90(image):
    return image.rotate(90)

def rotate_image_270(image):
    return image.rotate(270)

if __name__ == "__main__":
    input_folder = r"E:\My PHD\4-Cervical research paper\4-Datasets\AnnoCerv-main\Augmented and preprocessed\case 5-6 images"
    output_folder = r"E:\My PHD\4-Cervical research paper\4-Datasets\AnnoCerv-main\Augmented and preprocessed\case 5-6 images"  # Can be same as input
    total_target_images = 311

    image_augmentation(input_folder, output_folder, total_target_images)
```

**4. Augmentation-II (Random Variations - Brightness, Contrast, Translation, Rotation)**

```python
from PIL import Image, ImageEnhance
import os
import random

def image_augmentation(input_folder, output_folder, augment_count=4):
    # Create the output folder if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # List of augmentation functions to apply
    augmentation_functions = [
        adjust_brightness,
        adjust_contrast,
        translate_image,
        rotate_image
    ]

    # Loop through all images in the input folder
    for filename in os.listdir(input_folder):
        img_path = os.path.join(input_folder, filename)
        if not os.path.isfile(img_path):
            continue

        # Load the image using PIL
        img = Image.open(img_path)

        # Apply augmentations
        for i in range(augment_count):
            augmented_img = img.copy()
            augmentation_function = augmentation_functions[i]
            augmented_img = augmentation_function(augmented_img)

            # Save the augmented image to the output folder
            output_path = os.path.join(output_folder, f"{filename.split('.')[0]}_{i}new.jpg")
            augmented_img.save(output_path)

def adjust_brightness(image, factor_range=(0.7, 1.3)):
    factor = random.uniform(factor_range[0], factor_range[1])
    enhancer = ImageEnhance.Brightness(image)
    return enhancer.enhance(factor)

def adjust_contrast(image, factor_range=(0.5, 1.6)):
    factor = random.uniform(factor_range[0], factor_range[1])
    enhancer = ImageEnhance.Contrast(image)
    return enhancer.enhance(factor)

def translate_image(image):
    width, height = image.size
    shift_x = random.randint(-50, 50)
    shift_y = random.randint(-50, 50)
    return image.transform((width, height), Image.AFFINE, (1, 0, shift_x, 0, 1, shift_y))

def rotate_image(image):
    angle = random.randint(-150, 150)
    return image.rotate(angle)

if __name__ == "__main__":
    input_folder = r"E:\My PHD\4-Cervical research paper\4-Datasets\AnnoCerv-main\Augmented and preprocessed\Class3" # Replace this with your input folder containing images
    output_folder = r"E:\My PHD\4-Cervical research paper\4-Datasets\AnnoCerv-main\Augmented and preprocessed\Class3"   # Replace this with the output folder to save augmented images
    augment_count = 4  # Number of augmented images per original image this can be modified according to your need 
    image_augmentation(input_folder, output_folder, augment_count)
    
```
