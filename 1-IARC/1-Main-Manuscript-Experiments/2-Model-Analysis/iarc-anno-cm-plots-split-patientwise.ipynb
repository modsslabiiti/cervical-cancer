```python
import numpy as np
import matplotlib.pyplot as plt

# ================================
# Aggregated Confusion Matrices
# ================================

# --- IARC Dataset (you can replace with your own aggregated IARC matrix if needed) ---
cm_iarc = np.array([
    [207,  11,  8],
    [8, 51, 4],
    [ 6,  0, 81]
])

# --- AnnoCerv Dataset (Your final aggregated matrix) ---
cm_annocerv = np.array([
    [262,  23,  26],
    [ 16, 106,  2],
    [ 14,   2, 80]
])

class_labels_iarc = ["Type1", "Type2", "Type3"]
class_labels_annocerv = ["Category1", "Category2", "Category3"]


# ================================
# Function to Plot Confusion Matrix
# ================================

def plot_confusion_matrix(cm, class_names):
    plt.figure(figsize=(8, 7))
    
    im = plt.imshow(
        cm,
        cmap=plt.cm.Blues,
        interpolation='nearest',
        vmin=0,
        vmax=np.max(cm) * 1.2
    )

    # Colorbar with larger font
    cbar = plt.colorbar(im, fraction=0.046, pad=0.04)
    cbar.ax.tick_params(labelsize=14)   # <-- increase colorbar font size

    tick_marks = np.arange(len(class_names))
    plt.xticks(tick_marks, class_names, fontsize=14)
    
    ax = plt.gca()
    ax.set_yticks(tick_marks)
    ax.set_yticklabels(
        class_names,
        fontsize=14,
        rotation=90,
        va='center',
        ha='right'
    )

    # Write values inside cells
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(
                j, i, cm[i, j],
                ha="center",
                va="center",
                fontsize=18,
                color="black"
            )

    plt.xlabel("Predicted Label", fontsize=18)
    plt.ylabel("True Label", fontsize=18)

    plt.tight_layout()
    plt.grid(False)
    plt.show()





# ================================
# Plot Both Matrices
# ================================

plot_confusion_matrix(cm_iarc, class_labels_iarc)

plot_confusion_matrix(cm_annocerv, class_labels_annocerv)
```
